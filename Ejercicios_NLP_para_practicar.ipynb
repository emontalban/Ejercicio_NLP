# Completa los espacios vacios

```python
review = "I have been FLYING a lot lately and the Flights just keep getting DELAYED. Honestly, traveling for WORK gets exhausting with endless delays, but every trip teaches you something new!"

# Lowercase the review
lower_text = review.lower()
print("Todo a minusculas: " + lower_text + "\n")

# Tokenize the lower_text into words
tokens = word_tokenize(lower_text)
print(f'Texto separado en tokens {tokens} \n')

# Remove stop words and punctuation
stop_words = set(stopwords.words("english"))
clean_tokens = [palabra for palabra in tokens if palabra not in stop_words and palabra.isalpha()]

print(f'La limpieza de token: {clean_tokens}')
```

**Ejercicio 2:** El Limpiador de Signos (Regex básico)
Los signos de puntuación como !!!, ??? o ... suelen estorbar para analizar el significado básico.

**Reto: **Toma la frase: "¡Increíble proyecto..., me encanta la IA!!!" y elimina todos los signos de puntuación.

Pista: Puedes usar el módulo re (Regular Expressions) o el string punctuation de la librería string.

```python
# importada la libreria re

frase_IA = "¡Increíble proyecto..., me encanta la IA!!!"

#re.sub()? Busca lo que coincide con un patrón y lo sustituye en este caso por nada lo elimina
# ^ al principio del conjunto	Significa «todo lo que no sea.
#\w	Conserva caracteres de palabra: letras Unicode, números y _.
#\s	Conserva espacios y otros caracteres de separación.

frase_sin_puntuacion = re.sub(
    r"[^\w\s]",
    "",
    frase_IA
)

# Mostramos el resultado limpio.
print(frase_sin_puntuacion)
```

```python
# quitar signo de puntuacion con libreria string

frase_IA_libreria = "¡Increíble proyecto..., me encanta la IA con libreria!!!??"

# Reunimos los signos que queremos eliminar.
puntuacion = string.punctuation + "¡¿"

# Conservamos cada carácter que no sea un signo de puntuación.
frase_sin_puntuacion_II = "".join(
    caracter for caracter in frase_IA_libreria
    if caracter not in puntuacion
)

print(frase_sin_puntuacion_II)
```

**Ejercicio 3:** El Filtro de "Ruido" (Stop Words personalizadas)
A veces, las listas de stop words por defecto de NLTK no son suficientes.

**Reto:** Tienes este texto: "El usuario entró en la web y compró un producto ayer."

Reto: Crea tu propia lista de palabras prohibidas que incluya ["ayer", "entró", "en"] y límpialas del texto además de las que ya trae NLTK.

```python
texto = "El usuario entró en la web y compró un producto ayer."

# Convertimos el texto a minúsculas.
texto_minusculas = texto.lower()

# Separamos el texto en tokens.
tokens = word_tokenize(
    texto_minusculas,
    language="spanish"
)

# Cargamos las stopwords del español.
stop_words = set(stopwords.words("spanish"))

# Añadimos las palabras.
stop_words.update(["ayer", "entró", "en"])

# Conservamos las palabras que superan los dos filtros.
palabras_limpias = []

for palabra in tokens:
    if palabra not in stop_words and palabra.isalpha():
        palabras_limpias.append(palabra)

# Unimos las palabras para recuperar una frase.
texto_limpio = " ".join(palabras_limpias)

print(texto_limpio)
```

**Ejercicio 4:** El Unificador (Stemming)
Como mencionaste antes, queremos que "lento" y "lenta" sean tratados igual.

**Reto:** Investiga y usa el SnowballStemmer de nltk.stem.

Prueba: Aplica el stemmer a estas palabras: ["correr", "corriendo", "corrió", "corredor"]. ¿Qué resultado te da? ¿Logra extraer la "raíz"?

```python
stemmer = SnowballStemmer("spanish")

palabras = ["correr", "corriendo", "corrió", "corredor"]

for palabra in palabras:
    raiz = stemmer.stem(palabra)
    print(palabra, "→", raiz)
```

**Ejercicio 5:** El Limpiador de Emails (Extracción de datos)
A veces el texto viene con datos sensibles o irrelevantes que queremos quitar, como correos electrónicos.

**Reto: **Tienes el texto: "Contactame en juan.perez@email.com o en soporte@tienda.es para más info."

**Reto:** Usa una expresión regular (Regex) para sustituir cualquier email por la palabra [EMAIL].

Analogía: Es como cuando en una película censuran un nombre con un pitido; aquí queremos "censurar" los datos privados para que la IA no los aprenda.

```python
texto = "Contactame en juan.perez@email.com o en soporte@tienda.es para más info."

patron_email = r"[\w.+-]+@[\w.-]+\.[a-zA-Z]{2,}"

texto_sin_emails = re.sub(
    patron_email,
    "[EMAIL]",
    texto
)

print(texto_sin_emails)
```
